```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sentences=[
'I love reading books.',
'I love listening to music.',
'The sun rises in the east.',
'Birds are flying in the sky.',
'Python is easy to learn.',
'Artificial intelligence is transforming technology.',
'Neural networks learn from data.',
'Data science combines statistics and programming.',
'Students attend classes every day.',
'The weather is pleasant today.'
]
```

```python
voc_size=10000
```

```python
one_hot_repr=[one_hot(Words,voc_size) for Words in sentences]
one_hot_repr
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length=8
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
```

```python
embedded_docs #this make all sentence same
```

```python
#feature representation
dim=10
```

```python
model=Sequential()
model.add(Embedding(input_dim=voc_size,output_dim=dim,input_length=sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
model(embedded_docs)
model.summary()
```

```python
# model.predict(embedded_docs[0].shape)
# model.summary()
prediction = model.predict(embedded_docs[0:1])
print(prediction.shape)
```

```python
prediction = model.predict(embedded_docs[0:])

print(prediction.shape)
```
